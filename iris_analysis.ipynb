## Task 5: Filter and Analyze Data

```python
long_sepal = df[df["sepal_length"] > 6]
long_sepal
```

```python
df[df["petal_width"] > 1.5]
```

```python
df[df["species"] == "setosa"]
```

```python
df.groupby("species")["petal_length"].mean().idxmax()
```

```python
df.sort_values(by="sepal_length", ascending=False)
```

**Explanation:**
- 61 flowers have a sepal length greater than 6.
- 52 flowers have a petal width greater than 1.5.
- There are 50 setosa flowers.
- Virginica has the highest average petal length.
- After sorting, the longest sepal is 7.9 and belongs to a virginica flower.

## Task 6: Export Analysis Results

```python
import os
os.makedirs("output", exist_ok=True)
```

```python
species_summary = df.groupby("species").mean()
species_summary.to_csv("output/species_summary.csv")
species_summary
```

```python
long_sepal = df[df["sepal_length"] > 6]
long_sepal.to_csv("output/filtered_iris.csv", index=False)
long_sepal.shape
```

```python
os.listdir("output")
```

**Explanation:** The output folder was created. species_summary.csv contains the average measurements for each species, and filtered_iris.csv contains the 61 flowers with sepal length greater than 6. Both files were created successfully.

## Summary of Findings

- The dataset is clean: 150 rows, 5 columns, no missing values, and only 1 duplicate row.
- It is balanced, with 50 flowers in each of the 3 species.
- Petal measurements separate the species best: setosa has the smallest petals and virginica the largest.
- Virginica has the highest average petal length and dominates the long-sepal filter.
